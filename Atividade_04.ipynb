```python
temps = []
for i in range(7):
  tp = float(input("Digite aqui a temperatura : "))
  temps.append(tp)

soma = 0
for t in temps:
  soma += t
med = soma/7
ab = ac = ig = 0
for t in temps :
  if t < med:
    ab += 1
  elif t > med :
    ac += 1
  else:
    ig += 1
print(
      f"média : {med}"
      f"acima : {ac}, abaixo : {ab}, iguais : {ig}"
  )
```
